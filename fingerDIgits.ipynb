```python
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
(x_train, y_train), (x_test, y_test) =  mnist.load_data()
#plt.imshow(y_train[0])
#print(y_train[0])
print(x_train[0].shape)
```

```python
#x_train[0]
Nx_train = x_train / 255
Nx_test = x_test / 255

print(Nx_train[0])
```

```python
Cy_train = to_categorical(y_train, 10)
Cy_train

Cy_test = to_categorical(y_test, 10)
Cy_test
```

```python
sequential = Sequential([
    Flatten(input_shape = (28, 28)),
    Dense(128, activation = "relu"),
    Dense(64, activation = "relu"),
    Dense(10, activation = "softmax")
])
```

```python
sequential.compile(optimizer = "adam", loss = "categorical_crossentropy", metrics = ["accuracy"])
sequential.fit(Nx_train, Cy_train, epochs = 10, batch_size = 5, validation_split = 0.1, verbose = 1)
```

```python
s_loss, s_accuracy = sequential.evaluate(Nx_test, Cy_test, verbose = 0)
print("Neural Network accuracy for squential: ", s_accuracy)
```

```python
sequential.save("Sequential.h5")
```
