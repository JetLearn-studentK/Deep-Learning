```python
# Import libraries

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
#pip install tensorflow
print(tf.__version__)
```

```python
# Load the iris dataset
iris = load_iris()
X = iris.data # Features are : sepal length, sepal width, petal length, petal width
X
```

```python
Y = iris.target.reshape(-1, 1)
Y # Target: 0 = setosa, 1 = versicolor, 2 = virginica
```

```python
print(Y)
```

```python
encoder = OneHotEncoder(sparse_output = False)
y_encoder = encoder.fit_transform(Y)
print(y_encoder)
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y_encoder, test_size = 0.2, random_state = 42)
x_train_nn, x_test_nn, y_train_nn, y_test_nn = train_test_split(X, y_encoder, test_size = 0.2, random_state = 42)
```

```python
# Machine Learning Classifier

tree_model = DecisionTreeClassifier()
tree_model.fit(x_train, y_train)

y_pred_tree = tree_model.predict(x_test)
tree_accuracy = accuracy_score(y_test, y_pred_tree)
print("Decision Tree accuracy: ", tree_accuracy)
```

```python
# Neural network classifier
nn_model = Sequential([
    Dense(16, activation = "relu", input_shape = (4, )),
    Dense(16, activation = "relu"), 
    Dense(3, activation = "softmax")])


```

```python
nn_model.compile(optimizer = "adam", loss = "categorical_crossentropy", metrics = ["accuracy"])
nn_model.fit(x_train_nn, y_train_nn, epochs = 50, batch_size = 5, validation_split = 0.1, verbose = 1)
```

```python
# Evaluate the neural network model

nn_loss, nn_accuracy = nn_model.evaluate(x_test_nn, y_test_nn, verbose = 0)
print("Neural Network accuracy: ", nn_accuracy)
```
