```python
from tensorflow.keras.layers import Dense, SimpleRNN
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.utils import to_categorical
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np
```

```python
words = ""
phrase = "Good Morning how is your morning"
clean_phrase = phrase.lower()
words = clean_phrase.split(" ")
```

```python
print(words)
```

```python
words_set = set(words)
words_set = list(words_set)
dict1 = {}
dict2 = {}
for word in range(len(words_set)):
    dict1[words_set[word]] = word
    dict2[word] = words_set[word]
```

```python
print(f"{dict1}\n{dict2}")
```

```python
def convert_phrase(phraseP):
    list = []
    for word in phraseP:
        if word != "0":
            list.append(dict1[word.lower()])
    return list


converted_phrase = convert_phrase(words)
print(converted_phrase)
```

```python
sequence_length = 1
features = []
target = []
for i in range(0, (len(converted_phrase)-sequence_length)):
    slice = converted_phrase[i:i+sequence_length]
    features.append(slice)
    target.append(converted_phrase[i+sequence_length])
print(features)
print(target)
```

```python
baseArray = np.zeros(((len(converted_phrase)-sequence_length), 1, len(converted_phrase)))
baseArray
```

```python
for w in range(len(baseArray)):
    c = features[w][0]
    baseArray[w][0][c] = 1
print(baseArray)
```

```python
sequentialRNNwords = Sequential([
    SimpleRNN(128, input_shape = (sequence_length, len(words))),
    Dense(len(words), activation='sigmoid')
])
```

```python
sequentialRNNwords.compile(optimizer = "adam", metrics = ["accuracy"], loss = "sparse_categorical_crossentropy")
```

```python
expTarget = np.expand_dims(target, axis = (1))
```

```python
sequentialRNNwords.fit(baseArray, expTarget, epochs=50)
```

```python
prediction = dict2[np.argmax(sequentialRNNwords.predict(baseArray))]
print(prediction)
```
