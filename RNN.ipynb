```python
from tensorflow.keras.layers import Dense, SimpleRNN
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.utils import to_categorical
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np
```

```python
phrase = "Good Morning"
characters = list(set(phrase.lower()))
```

```python
characters
```

```python
dict1 = {}
dict2 = {}
for letter in range(len(characters)):
    dict1[characters[letter]] = letter
    dict2[letter] = characters[letter]
```

```python
print(dict1)
print(dict2)
```

```python
def convert_phrase(phraseP):
    list = []
    for letter in phraseP:
        if letter != "0":
            list.append(dict1[letter.lower()])
    return list


converted_phrase = convert_phrase(phrase)
print(converted_phrase)
```

```python
sequence_length = 4
```

```python
(len(converted_phrase)-sequence_length)+1
```

```python
sequence_length = 4
features = []
target = []
for i in range(0, (len(converted_phrase)-sequence_length)):
    slice = converted_phrase[i:i+sequence_length]
    features.append(slice)
    target.append(converted_phrase[i+sequence_length])
print(features)
print(target)
```

```python
baseArray = np.zeros((8, 4, 8))
baseArray[0]
```

```python
#baseArray[0][0][5] = 1
#print(baseArray)
```

```python
for b in range(len(baseArray)):
    for r in range(len(baseArray[b])):
        c = features[b][r]
        baseArray[b][r][c] = 1
print(baseArray)
```

```python
sequentialRNN = Sequential([
    SimpleRNN(128, input_shape = (sequence_length, len(characters))),
    Dense(len(characters), activation='sigmoid')
])
```

```python
sequentialRNN.compile(optimizer = "adam", metrics = ["accuracy"], loss = "sparse_categorical_crossentropy")
```

```python
expTarget = np.expand_dims(target, axis = (1))
```

```python
sequentialRNN.fit(baseArray, expTarget, epochs=50)
```
