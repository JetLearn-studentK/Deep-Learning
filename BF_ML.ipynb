```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression as lr
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_squared_error
```

```python
test = pd.read_csv("test_black_friday.csv")
train = pd.read_csv("train_black_friday.csv")

```

```python
x_train = train.drop(columns = "Purchase")
y_train = train["Purchase"]

x_test = train.drop(columns = "Purchase")
y_test = train["Purchase"]
```

```python
x_train = x_train.fillna(0)
y_train = y_train.fillna(0)
x_test = x_test.fillna(0)
y_test = y_test.fillna(0)
```

```python
le = LabelEncoder()
x_train["User_ID"] = le.fit_transform(x_train["User_ID"])
x_train["Product_ID"] = le.fit_transform(x_train["Product_ID"])
x_train["Gender"] = le.fit_transform(x_train["Gender"])
x_train["City_Category"] = le.fit_transform(x_train["City_Category"])
x_train["Age"] = le.fit_transform(x_train["Age"])
x_train["Stay_In_Current_City_Years"] = le.fit_transform(x_train["Stay_In_Current_City_Years"])


x_test["User_ID"] = le.fit_transform(x_test["User_ID"])
x_test["Product_ID"] = le.fit_transform(x_test["Product_ID"])
x_test["Gender"] = le.fit_transform(x_test["Gender"])
x_test["City_Category"] = le.fit_transform(x_test["City_Category"])
x_test["Age"] = le.fit_transform(x_test["Age"])
x_test["Stay_In_Current_City_Years"] = le.fit_transform(x_test["Stay_In_Current_City_Years"])
```

```python
lr_model = lr()
lr_model.fit(x_train, y_train)
```

```python
ytestpredict = lr_model.predict(x_test)
rootmeansquareerror = (np.sqrt(mean_squared_error(y_test,ytestpredict)))
print(rootmeansquareerror)
```
