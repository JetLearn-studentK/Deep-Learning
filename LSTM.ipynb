```python
import yfinance
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import numpy as np

apple = yfinance.download("AAPL", start = "2016-03-01", end = "2026-03-01")
apple
```

```python
apple["Close"]
```

```python
minMax = MinMaxScaler(feature_range = (0, 1))
CloseMinMax = minMax.fit_transform(apple["Close"])
CloseMinMax
```

```python
features = []
target = []

for i in range(len(CloseMinMax)//60):
    #tempList = []
    start = i * 60
    end = start + 60
    features.append(CloseMinMax[start:end])
    target.append(CloseMinMax[end:end+1])


#len(features)
#print(features[0][0])
#print(target[1])
print(target)
print(CloseMinMax[120])
# 1 to 60, 2 to 61 ... for features
# 61, 62 ... for target
```
