```python
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
test = pd.read_csv("test_black_friday.csv")
train = pd.read_csv("train_black_friday.csv")
#(x_train, y_train), (x_test, y_test) =  mnist.load_data()
```

```python
x_train = train.drop(columns = "Purchase")
y_train = train["Purchase"]

x_test = train.drop(columns = "Purchase")
y_test = train["Purchase"]
```

```python
#x_train.sample(5)
y_train.sample(5)
```

```python
x_train = x_train.fillna(0)
y_train = y_train.fillna(0)
x_test = x_test.fillna(0)
y_test = y_test.fillna(0)

#x_train.sample(10)
```

```python
x_train.info()
```

```python
y_train
```

```python

le = LabelEncoder()
x_train["User_ID"] = le.fit_transform(x_train["User_ID"])
x_train["Product_ID"] = le.fit_transform(x_train["Product_ID"])
x_train["Gender"] = le.fit_transform(x_train["Gender"])
x_train["City_Category"] = le.fit_transform(x_train["City_Category"])
x_train["Age"] = le.fit_transform(x_train["Age"])
x_train["Stay_In_Current_City_Years"] = le.fit_transform(x_train["Stay_In_Current_City_Years"])


x_test["User_ID"] = le.fit_transform(x_test["User_ID"])
x_test["Product_ID"] = le.fit_transform(x_test["Product_ID"])
x_test["Gender"] = le.fit_transform(x_test["Gender"])
x_test["City_Category"] = le.fit_transform(x_test["City_Category"])
x_test["Age"] = le.fit_transform(x_test["Age"])
x_test["Stay_In_Current_City_Years"] = le.fit_transform(x_test["Stay_In_Current_City_Years"])


```

```python
y_train
```

```python
SS = StandardScaler()
x_train = SS.fit_transform(x_train)
```

```python
sequential = Sequential([
    Dense(256, activation = "relu"),
    Dense(128, activation = "relu"),
    Dense(64, activation = "relu"),
    Dense(1)
])
```

```python
sequential.compile(optimizer = "adam", loss = "mean_squared_error", metrics = ["mean_squared_error"])
sequential.fit(x_train, y_train, epochs = 10, batch_size = 5, validation_split = 0.1, verbose = 1)
```

```python
sequential.save("sequentialHW.h5")
```

```python
from tensorflow.keras.models import load_model
sequentialNN = load_model("sequentialHW.h5")
```

```python
s_loss, s_accuracy = sequentialNN.evaluate(x_test, y_test, verbose = 0)
print("Neural Network accuracy for squential: ", s_accuracy)
```
